```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
m = 200 #Number of Samples
X = np.random.uniform(0, 5, m)
epsilon = np.random.normal(0, 1, m) #Noise
y = 3 + 4*X + epsilon
```

```python
#Plot X and Y

plt.figure(figsize = (7,5))
plt.scatter(X,y)
plt.xlabel("X")
plt.ylabel("y")
plt.show()
```

### Gradient Descent

```python
#Initialization
X_b = np.c_[np.ones(m), X]
theta = np.zeros(2) #Initial Theta Values [0,0]
eta = 0.05 #Learning Rate
n_iterations = 1000 #Number of Iterations
losses = [] #Storing losses for loss vs iterations plot
```

```python
#Implementing Gradient Descent
for i in range(n_iterations):
    y_pred = X_b.dot(theta) #Predicted value
    error = y_pred - y 
    mse = np.mean(error**2) #Mean Squared Error
    losses.append(mse)

    gradients = (2/m) * X_b.T.dot(error)
    theta = theta - eta*gradients #Next theta
```

```python
#Loss Vs Iterations Plot
plt.figure(figsize = (7,5))
plt.plot(losses)
plt.xlabel("Iterations")
plt.ylabel("MSE")
plt.title("Loss vs. Iterations")
plt.show()
```

```python
#Final Outputs of Gradient Descent
print("Final Intercept and Slope: ")
print("Theta 0: ", round(theta[0], 4))
print("Theta 1: ", round(theta[1], 4))
```

### Closed Form

```python
theta_normal = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)
```

```python
print("Final Intercept and Slope:")
print("Theta 0: ", round(theta_normal[0],4))
print("Theta 1: ", round(theta_normal[1],4))
```

### Plots of Gradient Descent vs Closed Form

```python
plt.figure(figsize = (7,5))
plt.scatter(X, y, label = "Data")
plt.plot(X, X_b.dot(theta), color = "red")
plt.xlabel("X")
plt.ylabel("y")
plt.title("Linear Regression - Gradient Descent")
plt.figure(figsize = (7,5))
plt.scatter(X, y)
plt.plot(X, X_b.dot(theta_normal), color="black")
plt.xlabel("X")
plt.ylabel("y")
plt.title("Linear Regression - Closed Form")
plt.show()
```

### Comparing Gradient Descent and Closed-Form

```python
print("Gradient Descent Intercept and Slope: ")
print("Theta 0: ", theta[0])
print("Theta 1: ", theta[1], end="\n")
print()
print("Normal Equation Intercept and Slope: ")
print("Theta 0: ", theta_normal[0])
print("Theta 1: ", theta_normal[1])
print()
print("Difference in Intercepts and Slopes: ")
print("Theta 0 Difference: ", abs(theta[0]-theta_normal[0]))
print("Theta 1 Difference: ", abs(theta[1]-theta_normal[1]))
```

### Did Gradient Descent converge to the same solution as the closed form?
As we can see from these conclusions, the difference between theta values from gradient descent and closed-form solution is neglegible. Hence, gradient descent did converge to the same solution as the closed form solution.
